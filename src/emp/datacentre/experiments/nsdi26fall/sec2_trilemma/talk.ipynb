```python
import numpy as np
import matplotlib.pyplot as plt

def radar_chart_empty(categories, *, rmin=0, rmax=4, top_scale=0.5):
    N = len(categories)
    angles = np.linspace(0, 2 * np.pi, N, endpoint=False).tolist()

    fig, ax = plt.subplots(figsize=(7, 7), subplot_kw=dict(polar=True))
    fig.patch.set_facecolor("#0F172A")
    ax.set_facecolor("#0F172A")

    ax.set_theta_offset(np.pi / 2)
    ax.set_theta_direction(-1)

    ax.set_ylim(rmin, rmax)

    ax.set_xticks(angles)
    ax.set_xticklabels([])
    ax.set_yticks([])
    ax.set_yticklabels([])

    ax.yaxis.grid(False)
    ax.xaxis.grid(False)
    ax.spines["polar"].set_visible(False)

    def scale_at_angle(theta):
        upper_amount = np.maximum(np.cos(theta), 0)
        return 1 - (1 - top_scale) * upper_amount

    dense_angles = np.linspace(0, 2 * np.pi, 500)

    # Rounded gridlines
    for r in [1, 2, 3, 4]:
        scaled_r = r * scale_at_angle(dense_angles)
        ax.plot(
            dense_angles,
            scaled_r,
            linewidth=3,
            alpha=0.7,
            linestyle="--",
            color="white",
        )

    # Spokes
    for a in angles:
        spoke_r = np.linspace(0, rmax, 100)
        scaled_spoke_r = spoke_r * scale_at_angle(a)

        ax.plot(
            np.full_like(spoke_r, a),
            scaled_spoke_r,
            linewidth=3,
            alpha=0.7,
            linestyle="--",
            color="white",
        )

    plt.tight_layout()
    return fig, ax


categories = ["Performance", "Deployability", "Fault Tolerance"]

fig, ax = radar_chart_empty(categories, rmin=0, rmax=4, top_scale=0.5)

plt.savefig("talk_trilemma_oppo.pdf", dpi=600, bbox_inches="tight")
# plt.show()
```
