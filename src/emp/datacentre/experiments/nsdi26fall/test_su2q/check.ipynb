```python
# calculate the average path length if uniform traffic
# compare su2 and su2q for DRing cluster_c -- this is what we have now
homedir = '/mnt/mirabelle/az6922_homedir/DRing/src/emp/datacentre/'
netpathfile = f'{homedir}evalnetpathfiles/netpath_dring_80_64_su2.np'
su2pwfileprefix = f'{homedir}rawpathweightfiles/pathweight_dring_2988_80_64_su2_cluster_c_load'
su2qpwfileprefix = f'{homedir}rawpathweightfiles/pathweight_dring_2988_80_64_su2q_cluster_c_load'
loadlist = range(5,51,5)
intervallist = range(48)
numsw = 80
```

```python
# read netpathfile
netpath = list()
for i in range(numsw):
    netpath.append(list())
    for j in range(numsw):
        netpath[i].append(list())
with open(netpathfile,'r') as f:
    lines = f.readlines()
    # produce
    fromsw = 0
    tosw = 0
    for line in lines:
        if "->" not in line:
            tokens = line.split()
            fromsw = int(tokens[0])
            tosw = int(tokens[1])
        else:
            path = [fromsw]
            tokens = line.split()
            for token in tokens:
                hops = token.split("->")
                path.append(int(hops[1]))
            netpath[fromsw][tosw].append(path)

for load in loadlist:
    for interval in intervallist:
        su2pwfile = f'{su2pwfileprefix}{load}_interval{interval}.var'
        su2qpwfile = f'{su2qpwfileprefix}{load}_interval{interval}.var'

        su2arr = list()
        for i in range(numsw):
            su2arr.append(list())
            for j in range(numsw):
                su2arr[i].append(list())
        with open(su2pwfile,'r') as f:
            lines = f.readlines()[1:]
            # 0,36,13,0.046875
            for line in lines:
                tokens = line.split(',')
                fromsw = int(tokens[0])
                tosw = int(tokens[1])
                pid = int(tokens[2])
                value = float(tokens[3])
                su2arr[fromsw][tosw].append([pid,value])

        su2qarr = list()
        for i in range(numsw):
            su2qarr.append(list())
            for j in range(numsw):
                su2qarr[i].append(list())
        with open(su2qpwfile,'r') as f:
            lines = f.readlines()[1:]
            for line in lines:
                tokens = line.split(',')
                fromsw = int(tokens[0])
                tosw = int(tokens[1])
                pid = int(tokens[2])
                value = float(tokens[3])
                su2qarr[fromsw][tosw].append([pid,value])

        su2plenlist = list()
        su2qplenlist = list()
        for fromsw in range(numsw):
            for tosw in range(numsw):
                if fromsw != tosw:
                    for pid,su2weight in su2arr[fromsw][tosw]:
                        path = netpath[fromsw][tosw][pid]
                        plen = len(path)-1
                        su2plenlist.append(plen*su2weight)
                    for pid,su2qweight in su2qarr[fromsw][tosw]:
                        path = netpath[fromsw][tosw][pid]
                        plen = len(path)-1
                        su2qplenlist.append(plen*su2qweight)

        print(f"load {load} interval {interval} su2 {sum(su2plenlist)/len(su2plenlist)} su2q {sum(su2qplenlist)/len(su2qplenlist)}")
```
