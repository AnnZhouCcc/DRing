For understanding the different ways of constructing a Dragonfly topology

```python
server_list = [100,3000]
w_list = range(2,21,2)
for total_num_server in server_list:
    print(f"total_num_server={total_num_server}")
    for w in w_list:
        g = w+1
        aplen = (w*w*g + 2*(w+g-3)*w*g + 3*(g-2)*(w-1)*w*g) / (w*g*(w*g-1))
        num_long_link = g*(g-1)
        num_short_link = w*(w-1)*g
        os = total_num_server/(w*g)
        print(f"w={w},os={os:.4f},aplen={aplen:.4f},nlonglink={num_long_link},nshortlink={num_short_link}")
```
