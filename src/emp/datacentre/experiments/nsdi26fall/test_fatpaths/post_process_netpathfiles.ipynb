```python
netpathfilefrom = "npfiles/netpath_dring_80_64_fatpaths_extra1_before.np"
netpathfileto = "npfiles/netpath_dring_80_64_fatpaths_extra1.np"
nswitches = 80
```

```python
with open(netpathfileto, 'w') as fw:
    with open(netpathfilefrom, 'r') as fr:
        lines = fr.readlines()
        iline = 0
        while iline < len(lines):
            line = lines[iline].strip()
            tokens = line.split()
            src = int(tokens[0])
            dst = int(tokens[1])
            npaths = int(tokens[2])
            if npaths==0:
                fw.write(f"{line}\n")
                iline += 1
            else:
                pathset = set()
                for p in range(npaths):
                    pathline = lines[iline + 1 + p].strip()
                    if pathline not in pathset:
                        pathset.add(pathline)
                fw.write(f"{src} {dst} {len(pathset)}\n")
                for pathline in pathset:
                    fw.write(f"{pathline}\n")
                iline += npaths+1
        
```
